```python
# Inizializza il client di Ollama
import ollama
```

```python
input_text = input("Enter text: ")
data = {"name": "Sara","age": 27,"city": "Bologna","profession": "Sviluppatrice di videogiochi",}

complention = ollama.chat(
    model='bonsai-2:9b',
    messages=[
        {
            "role": "system",
            "content": (
                "Sei uno scrittore professionista specializzato nella narrazione di dati. "
                "Trasforma il set di dati strutturati fornito in una descrizione biografica fluida, "
                "naturale e completa in italiano, collegando armoniosamente tutte le informazioni fornite."
            )
        },
        {
            "role": "user",
            "content": f"Ecco i dati da descrivere: {data}"
        }
    ],
    think=True,
    options={'num_ctx': 8192}
)

print(complention['message']['content'])
```
