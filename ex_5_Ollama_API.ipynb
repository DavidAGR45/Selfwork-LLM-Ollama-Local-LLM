```python
# Inizializza il client di Ollama
import ollama
```

```python
input_text = input("Enter text: ")


complention = ollama.chat(
    model='bonsai-2:9b',
    messages=[        
        {
            "role": "system",
            "content": (
                "Sei un insegnante esperto. Dato il testo fornito, genera 3 domande di comprensione "
                "educative per verificare l'apprendimento dei concetti chiave. "
                "Per ciascuna domanda fornisci sia il quesito che la relativa risposta corretta "
                "basata esclusivamente sul testo."
            )
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
    think=False,
    options={'num_ctx': 4096}
)

print(complention['message']['content'])
```
