**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python
# Inizializza il client di Ollama
import ollama
```

```python
input_text = input("Enter text: ")

complention = ollama.chat(
    model='bonsai-2:9b',
    messages=[  
        {
            "role": "system",
            "content": (
                "Sei un giornalista di sintesi. Riassumi l'articolo fornito nei suoi punti essenziali. "
                "Il riassunto deve essere composto da un singolo paragrafo e non deve tassativamente superare "
                "i 255 caratteri (spazi inclusi). Restituisci esclusivamente il testo del riassunto."
            )
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
    think=True,
    options={'num_ctx': 8192}
)

summary = complention['message']['content']

# Stampa il riassunto
print(summary)

# Conta ed evidenzia la lunghezza
print(f"\nConteggio caratteri: {len(summary)} / 255")

if len(summary) <= 255:
    print(" Rispetta il vincolo dei 255 caratteri!")
else:
    print(f" Attenzione: supera il limite di {len(summary) - 255} caratteri.")

```
