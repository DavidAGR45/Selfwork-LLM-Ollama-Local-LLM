**Traccia:** Scrivi un prompt che traduca un testo dall'italiano all'inglese e viceversa. Usa il prompt per tradurre una breve descrizione di un prodotto.

**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python
# Inizializza il client di Ollama
import ollama
```

```python
input_text = input("Enter text: ")

completion = ollama.chat(
    model='bonsai-2:9b',
    messages=[
        {
            'role': 'system',
            'content': (
                'Sei un traduttore professionista specializzato in schede prodotto. '
                'Traduci il testo fornito in modo naturale e fluido: se il testo e in italiano, '
                'traducilo in inglese; se e in inglese, traducilo in italiano. '
                'Restituisci esclusivamente il testo tradotto, senza introduzioni o commenti.'
            )
        },
        {
            'role': 'user',
            'content': input_text
        }
    ],
    think=False,
    options={'num_ctx': 4096}
)

print('\nTraduzione:')
print(completion['message']['content'])
```
